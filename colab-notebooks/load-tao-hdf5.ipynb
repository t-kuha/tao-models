```python
# import os
# os.environ["TF_USE_LEGACY_KERAS"] = "1"

import json
import tensorflow as tf

print(f'{tf.__version__ = :}')
print(f'CPUs: {tf.config.list_physical_devices("CPU")}')
print(f'GPUs: {tf.config.list_physical_devices("GPU")}')
# print(tf.keras.version())  # Non-legacy Keras
```

```python
with open('classmap.json', 'r') as f:
    cls = json.load(f)

labels = [k for k, v in cls.items()]
```

```python
# tf.keras.backend.set_image_data_format('channels_first')
# tf.keras.backend.set_image_data_format('channels_last')

model = tf.keras.models.load_model("resnet_18.hdf5", compile=False)
# model.summary()
```

```python
if True:
    img = tf.keras.preprocessing.image.img_to_array(
        tf.keras.preprocessing.image.load_img('ffae2f74dd51806b.jpg', target_size=(224, 224))
    )
else:
    # dummy input
    img = tf.ones((224, 224, 3), dtype=tf.float32)

# apply pre-processing
preprocessed = tf.keras.applications.resnet50.preprocess_input(img)
preprocessed = tf.transpose(preprocessed, [2, 0, 1])

# run inference
pred = model.predict(tf.expand_dims(preprocessed, 0))

# top-5
_, top5 = tf.math.top_k(pred, k=5, sorted=True)
for i, t in enumerate(top5[0]):
    print(f'{i}: {labels[t]}')

# model.save('resnet_18.keras')
```
